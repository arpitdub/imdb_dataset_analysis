```python
#NUMPY
# Q -- What is the average duration of all movies?

np.mean(df['Duration_minutes'])

# Statistical Aggregation (Mean) , NumPy mean function
```

```python
# Q -- Distribution of ratings using NumPy

ratings = df['Total_reviews'].dropna().values   #removes Nan values

print("Mean:", np.mean(ratings)) 
print("Median:", np.median(ratings)) 
print("Std Dev:", np.std(ratings))
```

```python
# Q -- label based data selection 
# as it will show copy do not chanhe the original

df.loc[(df['Year'] > 2000) & (df['Imdb_rating'] > 9.0)].sort_values(by='Year', ascending=False).head(5)

# Select specific rows and columns together
```

```python
# Q -- Which decade had the most movies?

df["Decade"] = (df["Year"]//10)*10   #2004 // 10 = 200 , 200*10 = 2000 Year Decade
print(df["Decade"].value_counts().sort_index())

#decade column will added
#Shows cinema boom decades.
```

```python
# Q -- Total movies released per year

df['Year'].value_counts().sort_index()

# -- Frequency Counting & Sorting Operation
```

```python
# Q -- Longest Movie Duration by Each Country in descending 0rder
#loc Label based With Row and Column 

Longmov = df.loc[
               df.groupby('Country')['Duration_minutes'].idxmax(),
                ['Country', 'Movie_title', 'Duration_minutes']].sort_values(by='Duration_minutes', ascending=False)

Longmov = Longmov.reset_index(drop=True)

print(Longmov)
```
